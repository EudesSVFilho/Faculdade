```python
for clf_name, clf, clf_params in cluster_methods:

    par = list(ParameterGrid(clf_params))

    best_estimator = clf
    best_params = par[0]

    best_score = 1e12

    for i in par:
        m = clf.set_params(**i)

        try:
            y_pred = m.fit_predict(x_scaler)
        except Exception as e:
            print(f"Pulando parametros {i} para {clf_name}: erro no fit_predict -> {e}")
            continue

        labels = set(y_pred)
        labels_no_noise = labels - {-1} if -1 in labels else labels
        n_labels = len(labels_no_noise)

        if n_labels < 2:
            print(f"Pulando parametros {i} para {clf_name}: apenas {n_labels} cluster(es) válido(s)")
            continue

        mask = np.array([lbl in labels_no_noise for lbl in y_pred])
        X_for_db = x_scaler[mask]
        y_for_db = np.array(y_pred)[mask]

        try:
            db = metrics.davies_bouldin_score(X_for_db, y_for_db)
        except Exception as e:
            print(f"Pulando parametros {i} para {clf_name}: erro ao calcular DB -> {e}")
            continue

        if db < best_score:
            best_score = db
            best_params = i

    print("Melhor score: "+str(best_score)+" Melhores parâmetros: "+str(best_params))
    clf = clf.set_params(**best_params)


    y_pred = clf.fit_predict(x_scaler)


    n_cluster = len(set(y_pred))


    col = ['r','b','g','m','y','c']
    clabel = [int(x) for x in y_pred]
    for i in range(len(clabel)):
      k = clabel[i]
      if k == -1:
          clabel[i] = 'k'
      else:
          clabel[i] = col[k % len(col)]

    mar = ['o','^','*','x','+']
    mlabel = [int(x) for x in y_pred]
    for i in range(len(mlabel)):
      l = mlabel[i]
      if l == -1:
          mlabel[i] = 'H'
      else:
          mlabel[i] = mar[l % len(mar)]

    pca = PCA(n_components = 2)

    cluster_reduced_data = pca.fit_transform(x_scaler)

    fig = plt.figure()
    x1 = cluster_reduced_data[:,0]; y1 = cluster_reduced_data[:,1]

    x1 = (x1 - x1.min()) /  (x1.max() - x1.min())
    y1 = (y1 - y1.min()) /  (y1.max() - y1.min())


    for i in range(len(x1)):
      plt.scatter(x1[i], y1[i], marker=mlabel[i], s=100, c=clabel[i])
      plt.text(x1[i] + 0.02, y1[i] + 0.00, '%s' % y[i], ha='left', va='center', fontsize=10)
    plt.xlim(-0.1, +1.4); plt.ylim(-0.1, +1.1)
    plt.xlabel('Componente 1 (explica '+str("{:.3f}".format(100*pca.explained_variance_ratio_[0]))+'% de variabilidade)', size='large')
    plt.ylabel('Componente 2 (explica '+str("{:.3f}".format(100*pca.explained_variance_ratio_[1]))+'% de variabilidade)', size='large')
    plt.title(clf_name)
    plt.savefig(clf_name + '.png', bbox_inches='tight')
    plt.show()
```
